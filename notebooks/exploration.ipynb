# Exploration Notebook

Notebook untuk eksplorasi data struktur/BOQ sebelum buat dashboard.

```python
import pandas as pd
import numpy as np
import json
from pathlib import Path

# Load JSON yang dihasilkan excel_to_json.py
json_path = Path('../data/sample_structure.json')
if json_path.exists():
    with open(json_path) as f:
        data = json.load(f)
    print(f'Loaded {len(data)} sheets')
    for name, sheet in data.items():
        df = pd.DataFrame(sheet['data'])
        print(f'\n{name}: {df.shape[0]} rows × {df.shape[1]} cols')
        print(df.head())
else:
    print('JSON not found. Jalankan excel_to_json.py dulu.')
```
